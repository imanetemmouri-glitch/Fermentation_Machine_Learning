```python
import os
import pandas as pd
import zipfile

# Récupérer les fichiers de 'uploaded' si disponible, sinon scanner '/content'
try:
  files_to_process = list(uploaded.keys())
except NameError:
  files_to_process = [f for f in os.listdir('/content') if f.endswith(('.zip', '.xlsx', '.xls'))]

if not files_to_process:
  print("Aucun fichier trouvé. Veuillez réexécuter la cellule de téléchargement (upload) ou déposer vos fichiers dans l'onglet de gauche.")

for f in files_to_process:
  file_path = os.path.join('/content', f) if not os.path.isabs(f) else f
  if not os.path.exists(file_path):
    continue

  if f.endswith('.zip'):
    with zipfile.ZipFile(file_path, 'r') as zip_ref:
      zip_ref.extractall('/content')
      for file_name in zip_ref.namelist():
        if file_name.endswith(('.xlsx', '.xls')):
          extracted_path = os.path.join('/content', file_name)
          df = pd.read_excel(extracted_path)
          print("\nfichier extrait :", file_name)
          print("colonnes :", df.columns.tolist())
          display(df.head(3))
  elif f.endswith(('.xlsx', '.xls')):
    df = pd.read_excel(file_path)
    print("\nfichier :", f)
    print("colonnes :", df.columns.tolist())
    display(df.head(3))
```

```python
import os
import pandas as pd
import zipfile

# Trouver tous les fichiers Excel directs ou extraits dans /content
excel_files = []
for root, dirs, files in os.walk('/content'):
  for file in files:
    if file.endswith(('.xlsx', '.xls')):
      excel_files.append(os.path.join(root, file))

if not excel_files:
  # Si aucun fichier Excel n'est encore extrait, on regarde si un ZIP est présent pour l'extraire
  zip_files = [f for f in os.listdir('/content') if f.endswith('.zip')]
  for z in zip_files:
    with zipfile.ZipFile(os.path.join('/content', z), 'r') as zip_ref:
      zip_ref.extractall('/content')
  # Re-recherche des fichiers Excel après extraction
  for root, dirs, files in os.walk('/content'):
    for file in files:
      if file.endswith(('.xlsx', '.xls')):
        excel_files.append(os.path.join(root, file))

# Affichage des feuilles pour chaque fichier Excel trouvé
if excel_files:
  for file_path in excel_files:
    rel_path = os.path.relpath(file_path, '/content')
    print("Nom du fichier :", rel_path)
    print("feuilles disponibles :", pd.ExcelFile(file_path).sheet_names)
    print("-" * 40)
else:
  print("Aucun fichier Excel ou archive ZIP trouvé. Veuillez réexécuter la cellule d'importation (upload).")
```
